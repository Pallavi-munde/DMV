# Experiment 6: Data Aggregation and Visualization of Sales Performance by Region

```python
# ===========================================================
# DATA AGGREGATION
# ANALYZING SALES PERFORMANCE BY REGION
# ===========================================================
import matplotlib.pyplot as plt

import pandas as pd

# Try several public copies of the Superstore dataset; fall back to manual upload.
urls = [
    "https://raw.githubusercontent.com/Victory-Data/Superstore-Sales-Dataset/main/Sample%20-%20Superstore.csv",
    "https://raw.githubusercontent.com/Victory-Data/Superstore-Sales-Dataset/master/Sample%20-%20Superstore.csv",
    "https://raw.githubusercontent.com/Victory-Data/Superstore-Sales-Dataset/main/Superstore.csv",
    "https://raw.githubusercontent.com/Victory-Data/Superstore-Sales-Dataset/main/train.csv",
]

df = None
for u in urls:
    try:
        df = pd.read_csv(u, encoding="latin1")
        print("Loaded from:", u)
        break
    except Exception:
        continue

if df is None:
    from google.colab import files
    print("Online copy not reachable. Please upload 'Sample - Superstore.csv' (from Kaggle).")
    up = files.upload()
    df = pd.read_csv(list(up.keys())[0], encoding="latin1")

print("Dataset Loaded Successfully!\n")
```

```python
# 2. Display first 5 records
print("First 5 Records:")
display(df.head())

# 3. Dataset information
print("\nDataset Shape:", df.shape)
print("\nRequired Columns:")
display(df[["Order Date", "Region", "Category", "Quantity", "Sales"]].head())

# 4. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 5. Remove missing and duplicate records
df = df.dropna()
df = df.drop_duplicates()

# Convert date
df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")
print("\nData Cleaning Completed!")
```

```python
# ===========================================================
# DATA AGGREGATION
# ===========================================================
# 6. Region-wise total sales
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
print("\nRegion-wise Total Sales:")
print(region_sales)

# 7. Region-wise quantity sold
region_quantity = df.groupby("Region")["Quantity"].sum().sort_values(ascending=False)
print("\nRegion-wise Quantity Sold:")
print(region_quantity)

# 8. Region-wise average sales
region_avg = df.groupby("Region")["Sales"].mean().sort_values(ascending=False).round(2)
print("\nRegion-wise Average Sales:")
print(region_avg)

# 9. Region-wise summary table
region_summary = df.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Transactions=("Sales", "count"),
).sort_values("Total_Sales", ascending=False)
print("\nRegion-wise Sales Summary:")
display(region_summary)

# 10. Top performing region
top_region = region_sales.idxmax()
print("\nTop Performing Region:", top_region)
print("Total Sales:", round(region_sales.max(), 2))

# 11. Category-wise sales
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
print("\nCategory-wise Sales:")
print(category_sales)

# Region x Category (stacked comparison)
region_category = df.groupby(["Region", "Category"])["Sales"].sum().unstack()
print("\nSales by Region and Category:")
display(region_category)
```

```python
# 12. Region-wise sales chart
region_sales.plot(kind="bar")
plt.title("Sales Performance by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Pie chart: sales distribution by region
plt.pie(region_sales, labels=region_sales.index, autopct="%1.1f%%")
plt.title("Sales Distribution by Region")
plt.show()

# 13. Region-wise quantity chart
region_quantity.plot(kind="bar")
plt.title("Quantity Sold by Region")
plt.xlabel("Region")
plt.ylabel("Quantity Sold")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# 14. Category-wise sales chart
category_sales.plot(kind="bar")
plt.title("Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Stacked bar: region vs category
region_category.plot(kind="bar", stacked=True)
plt.title("Sales by Region and Product Category")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# 15. Save aggregated result
region_summary.to_csv("Region_Sales_Summary.csv")
print("\nRegion-wise sales summary saved successfully!")
print("\nData Aggregation Completed Successfully!")
```
