# Experiment 5: Data Visualization using Matplotlib - AQI Trends

Note: the pollutant columns (PM2.5, PM10, CO, AQI) are generated randomly for demonstration (seed fixed), exactly as in the original practical.

```python
# ===========================================================
# DATA VISUALIZATION USING MATPLOTLIB
# ANALYZING AIR QUALITY TRENDS
# ===========================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Load dataset from online source (used for its Date column)
url = "https://raw.githubusercontent.com/plotly/datasets/master/2016-weather-data-seattle.csv"
df = pd.read_csv(url)
print("Dataset Loaded Successfully!\n")

# 2. Display first 5 rows
print("First 5 Records:")
display(df.head())
```

```python
# Create required pollutant columns for demonstration
np.random.seed(10)
n = len(df)

df["PM2.5"] = np.random.randint(20, 150, n)
df["PM10"] = np.random.randint(30, 200, n)
df["CO"] = np.random.uniform(0.5, 5, n).round(2)
df["AQI"] = np.random.randint(50, 300, n)

# 3. Convert Date column
df["Date"] = pd.to_datetime(df["Date"])

# Keep required columns
air_data = df[["Date", "PM2.5", "PM10", "CO", "AQI"]].copy()
print("\nAir Quality Data:")
display(air_data.head())

# 4. Dataset information
print("\nDataset Shape:", air_data.shape)
print("\nMissing Values:")
print(air_data.isnull().sum())

# 5. Remove missing values
air_data = air_data.dropna()
print("\nData Cleaning Completed!")

# 6. Statistical summary
print("\nStatistical Summary:")
display(air_data.describe())
```

```python
# 7. Line plots: AQI, PM2.5, PM10 over time
for col in ["AQI", "PM2.5", "PM10"]:
    plt.figure(figsize=(10, 4))
    plt.plot(air_data["Date"], air_data[col])
    plt.title(f"{col} Trend Over Time" if col == "AQI" else f"{col} Levels Over Time")
    plt.xlabel("Date")
    plt.ylabel(col)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
# 8. Bar plot: average pollutant levels
avg = air_data[["PM2.5", "PM10", "CO"]].mean()
plt.bar(avg.index, avg.values)
plt.title("Average Pollutant Levels")
plt.xlabel("Pollutants")
plt.ylabel("Average Level")
plt.show()

# 9. Histogram: AQI distribution
plt.hist(air_data["AQI"], bins=20)
plt.title("AQI Distribution")
plt.xlabel("AQI")
plt.ylabel("Frequency")
plt.show()

# 10. Box plot: distribution & outliers
plt.boxplot([air_data["PM2.5"], air_data["PM10"], air_data["AQI"]],
            labels=["PM2.5", "PM10", "AQI"])
plt.title("Box Plot of Pollutants and AQI")
plt.show()

# 11. Scatter plot: PM2.5 vs AQI
plt.scatter(air_data["PM2.5"], air_data["AQI"], alpha=0.5)
plt.title("PM2.5 vs AQI")
plt.xlabel("PM2.5")
plt.ylabel("AQI")
plt.show()

# Save dataset
air_data.to_csv("City_Air_Quality.csv", index=False)
print("\nDataset saved as City_Air_Quality.csv")
print("\nAir Quality Visualization Completed Successfully!")
```
