# Experiment 4: Data Wrangling on Real Estate Market (Boston Housing)

```python
# ===========================================================
# DATA WRANGLING ON REAL ESTATE MARKET DATASET
# ===========================================================
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load Real Estate dataset from online source
url = "https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv"
df = pd.read_csv(url)
print("Dataset Loaded Successfully!\n")

# 2. Display first 5 records
print("First 5 Records:")
display(df.head())

# 3. Dataset information
print("\nDataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
print("\nDataset Information:")
df.info()

# 4. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 5. Check duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
# 6. Remove duplicates
df = df.drop_duplicates()

# 7. Handle missing values (median imputation)
numeric_columns = df.select_dtypes(include="number").columns
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

# 8. Reset index
df = df.reset_index(drop=True)
print("\nData Wrangling Completed!")

# 9. Statistical summary
print("\nStatistical Summary:")
display(df.describe())
```

```python
# 10. Rename important columns for easy understanding
df = df.rename(columns={
    "rm": "Rooms",
    "medv": "House_Price",
    "age": "Property_Age",
    "crim": "Crime_Rate",
    "tax": "Tax",
})
print("\nColumns After Renaming:")
print(df.columns.tolist())

# 11. Basic price statistics
print("\nAverage House Price:")
print(round(df["House_Price"].mean(), 2))
print("\nMaximum House Price:")
print(df["House_Price"].max())
print("\nMinimum House Price:")
print(df["House_Price"].min())

# 12. Data transformation
print("\nData After Transformation:")
display(df.head())
```

```python
# 13. Feature engineering - price categories
df["Price_Category"] = pd.cut(
    df["House_Price"],
    bins=[0, 20, 35, 100],
    labels=["Low", "Medium", "High"]
)
print("\nPrice Categories:")
print(df["Price_Category"].value_counts())

# 14. Average price by number of rooms (first 10 groups)
print("\nAverage Price by Number of Rooms:")
print(df.groupby("Rooms")["House_Price"].mean().round(2).head(10))
```

```python
# 15. Distribution of house prices
plt.hist(df["House_Price"], bins=20)
plt.title("Distribution of House Prices")
plt.xlabel("House Price")
plt.ylabel("Frequency")
plt.show()

# 16. Rooms vs House Price
plt.scatter(df["Rooms"], df["House_Price"])
plt.title("Rooms vs House Price")
plt.xlabel("Average Number of Rooms")
plt.ylabel("House Price")
plt.show()

# 17. Property Age vs House Price
plt.scatter(df["Property_Age"], df["House_Price"])
plt.title("Property Age vs House Price")
plt.xlabel("Property Age")
plt.ylabel("House Price")
plt.show()
```

```python
# 19. Save wrangled dataset
df.to_csv("RealEstate_Prices_Cleaned.csv", index=False)
print("\nCleaned dataset saved as RealEstate_Prices_Cleaned.csv")

# 20. Final dataset
print("\nFinal Wrangled Dataset:")
display(df.head())
print("\nData Wrangling and Analysis Completed Successfully!")
```
