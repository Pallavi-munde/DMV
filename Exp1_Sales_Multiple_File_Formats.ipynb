```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import pandas as pd

# Try several public copies of the Superstore dataset; fall back to manual upload.
urls = [
    "https://raw.githubusercontent.com/Victory-Data/Superstore-Sales-Dataset/main/Sample%20-%20Superstore.csv",
    "https://raw.githubusercontent.com/Victory-Data/Superstore-Sales-Dataset/master/Sample%20-%20Superstore.csv",
    "https://raw.githubusercontent.com/Victory-Data/Superstore-Sales-Dataset/main/Superstore.csv",
    "https://raw.githubusercontent.com/Victory-Data/Superstore-Sales-Dataset/main/train.csv",
]

df = None
for u in urls:
    try:
        df = pd.read_csv(u, encoding="latin1")
        print("Loaded from:", u)
        break
    except Exception:
        continue

if df is None:
    from google.colab import files
    print("Online copy not reachable. Please upload 'Sample - Superstore.csv' (from Kaggle).")
    up = files.upload()
    df = pd.read_csv(list(up.keys())[0], encoding="latin1")

print("Dataset Loaded Successfully!\n")

print("Original Data:")
display(df.head())
```

```python
# Save data in different formats
df.to_csv("sales.csv", index=False)
df.to_excel("sales.xlsx", index=False)
df.to_json("sales.json", orient="records")

# Load different file formats
csv_data = pd.read_csv("sales.csv")
excel_data = pd.read_excel("sales.xlsx")
json_data = pd.read_json("sales.json")

# Combine data
data = pd.concat([csv_data, excel_data, json_data])

# Remove duplicates and missing values
data = data.drop_duplicates()
data = data.dropna()

print("\nCleaned Data:")
display(data.head())
```

```python
# Basic Analysis
print("Total Sales:", data["Sales"].sum())
print("Average Sales:", data["Sales"].mean())
print("Total Profit:", data["Profit"].sum())

# Category-wise sales
category_sales = data.groupby("Category")["Sales"].sum()
print("\nCategory-wise Sales:")
print(category_sales)
```

```python
# Graph
category_sales.plot(kind="bar")
plt.title("Category-wise Sales")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()

# Save final cleaned data
data.to_csv("cleaned_sales.csv", index=False)
print("\nAnalysis Completed Successfully!")
```
